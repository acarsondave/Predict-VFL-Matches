```python
import re
import time
import pandas
import schedule
import threading
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions
from selenium.common.exceptions import NoSuchElementException
from selenium.common.exceptions import StaleElementReferenceException

options = Options()
options.headless = True
options.page_load_strategy = 'eager'
driver = webdriver.Firefox(options=options, executable_path=r'/usr/bin/geckodriver')
driver.implicitly_wait(30) # seconds 
base_url = "https://rgs.betradar.com/ls/livescore/?/bet9ja/en/page/rgs_vfl/?key=pK9saJZcyZRVRgZ9&bk=27&do=dec&currency=ngn"
driver.get(base_url)
```

```python
def under35():
    a = 1
    b = 1
    occ_two = 0
    win_count = "0"
    loss_count = "0"
    soup = BeautifulSoup(driver.page_source, "html.parser")
    driver.switch_to.frame("vflframe")
    for items in range(0,30):
        matchday = "matchday%s" % str(items+1)
        driver.find_elements_by_id(matchday)[0].click()
        time.sleep(3)
        driver.switch_to.default_content()
        soup = BeautifulSoup(driver.page_source, "html.parser")
        mo = soup.find_all("div", {"class":"matchlist-market-name-text"})
        odd = soup.find_all("div",{"class":"matchlist-betting-options"})
        win = soup.find_all("label",{"class":"wonOdd"})
        lose = soup.find_all("label",{"class":"btnOdd"})
        idx = 0
        for items,odds,wins in zip(mo,odd,win):
            if " Over/under (3.5) " in items:
                overodds = odds.text.split("u")[0].replace(" ","")
                underodds = odds.text.split("u")[1].replace(" ","")
                wins = wins.text.replace(" ","")
                if "1.40" in underodds:
                    occ_two = occ_two + 1
                    print("%s\nWin: %s\nOverodd:%s\nOver 3.5\nWinCount: %s\nIndex no: %s\n" %(matchday,wins,overodds,a,idx))
                    if "1.40" in wins:
                        a = a + 1
            idx = idx + 1
        driver.switch_to.frame("vflframe")
    f = open("occur_35.txt", "w")
    f.write(str(occ_two))
    driver.switch_to.default_content()
```
